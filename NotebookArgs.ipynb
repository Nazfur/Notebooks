```python
def f(a):	    	
	a = 99 			

b = 88		
f(b)				
print("b = ", b)	# b no cambia porque se pasa como parametro
```

```python
def changer(a, b):	    
		a = 2			
		b[0] = 'spam' 	
					
x = 1
l = [1, 2]
changer(x, l)        # Si se pone solo l se cambia su valor!
print("l = ", l)        # ['spam', 2]  # x is unchanged, l is different!
assert x == 1, "x es 1"
```

```python
l = [1, 2]
changer(x, l[:]) 	# l[:] crea una copia para que no se modificque ese valor

print("l = ", l) 
```

```python
l = [1, 2]
def changer(a, b):
    b = b[:]		# Copiamos el valor de la lista
    a = 2
    b[0] = 'spam'
changer(x, l)
assert l == [1, 2], "l es una invariante" 
print("l = ", l) 
```

## Orden

```python
def f(a, b, c): 
    print(a, b, c)

f(c=3, b=2, a=1)    # 1 2 3
```

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)


f(1)        # 1 2 3
f(a=1)      # 1 2 3
```

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)
    
f(1, 4)     # 1 4 3
f(1, 4, 5)	# Override defaults: 1 4 5

f(1, c=6)	# Choose defaults: 1 2 6
1 4 3
1 4 5
1 2 6
```

```python
def func(a, b, c, d): 
    print(a, b, c, d)
args = {'a': 1, 'b': 2, 'c': 3}
args['d'] = 4
func(**args)    # Same as func(a=1, b=2, c=3, d=4)
func(*args)

func(*(1, 2), **{'d': 4, 'c': 3})       # Same as func(1, 2, d=4, c=3)
# 1 2 3 4
func(1, *(2, 3), **{'d': 4})            # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, c=3, *(2,), **{'d': 4})         # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
func(1, *(2, 3), d=4)                   # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, *(2,), c=3, **{'d':4})      # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
```

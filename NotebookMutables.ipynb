## Caso de las listas
---

```python
a = [1, 2]
b = a
a = [0, 0]

print(b)

a = [1, 2]
b = a
a.pop(0)
print(b)

L = [2, 3, 4]	    # A mutable object
M = L			    # Make a reference to the same object
L[0] = 24 		    # An in-place change

print("L = ", L)    # [24, 3, 4]
print("M = ", M)    # [24, 3, 4]
```

```python

L = [2, 3, 4]
M = L[:]		    # Make a copy of L (or list(L), copy.copy(L), etc.)
L[0] = 24

print("L = ", L)    # [24, 3, 4]

print("M = ", M)    # [2, 3, 4]  M is not changed
```

```python
L = [1, 2, 3]
M = L 			    # M and L reference the same object

print(L == M)			    # True: Same values => operador igualdad de valores
print(L is M)
```

### Cache de valores
Para dos valores pequeños se tiende a reusar la misma memoria
---

```python
X = 42
Y = 42		    # Should be two different objects

X  == Y         # True
X is Y 		    # True: Same object anyhow: caching at work!!
```
